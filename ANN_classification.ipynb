```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("DateFruit_Dataset.csv")
```

```python
df.isnull().sum()
```

```python
X=df.drop("Class",axis=1)
y=df["Class"]
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
y=le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(
    X,y,test_size=42,random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
X_scaled_train=scaler.fit_transform(X_train)
X_scaled_test=scaler.transform(X_test)
```

```python
import torch
import torch.nn as nn

X_tensor_train=torch.tensor(X_scaled_train,dtype=torch.float32)
y_tensor_train=torch.tensor(y_train,dtype=torch.long)

X_tensor_test=torch.tensor(X_scaled_test,dtype=torch.float32)
y_tensor_test=torch.tensor(y_test,dtype=torch.long)
```

```python
from torch.utils.data import TensorDataset,DataLoader
train_dataset=TensorDataset(X_tensor_train,y_tensor_train)
test_dataset=TensorDataset(X_tensor_test,y_tensor_test)
```

```python
from torch.utils.data import TensorDataset,DataLoader
train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True)
test_loader=DataLoader(test_dataset,batch_size=32)
```

```python
class ANN(nn.Module):
    def __init__(self):
        super(ANN,self).__init__()
        
        self.model=nn.Sequential(

        nn.Linear(X_train.shape[1],64),
        nn.ReLU(),
        nn.Linear(64,64),
        nn.ReLU(),
        nn.Linear(64,7)    
    )

    def forward(self,x):
        return self.model(x)
    
```

```python
import torch.optim as optim

model =ANN()

criteria=nn.CrossEntropyLoss()
optimizer=optim.Adam(model.parameters())
```

```python
epochs=100

for epoch in range(epochs):
    model.train

    running_loss=0.0

    for xb,yb in train_loader:
        optimizer.zero_grad()
        
        outputs=model(xb)
        loss=criteria(outputs,yb)
        loss.backward()
        optimizer.step()


        running_loss+=loss.item()

    train_loss = running_loss / len(train_loader)
    print(f"epoch={epoch+1}/{epochs},loss={train_loss}")
```

```python
#evaluate
model.eval()
total=0
correct=0

with torch.no_grad():
    for xb,yb in test_loader:
        outputs=model(xb)
        _,predicted=torch.max(outputs,1)

        correct+=(predicted==yb).sum().item()
        total+=yb.size(0)

print("total:",total)
print("correct:",correct)


print("accuracy:",correct/total*100)
```
